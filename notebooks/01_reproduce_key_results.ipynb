# Reproduce key PARAMO public results
This notebook reads only the curated public CSV files included in the repository. It does not require the private optimization model.

```python
from pathlib import Path
import pandas as pd
import matplotlib.pyplot as plt

ROOT = Path.cwd().parent if Path.cwd().name == 'notebooks' else Path.cwd()
summary = pd.read_csv(ROOT / 'data/national/scenario_summary.csv')
summary[['Hydrology','Scenario','MeanTotalCost_MUSD','MeanCO2_kt','MeanENS_GWh']]
```

```python
fig, ax = plt.subplots(figsize=(7, 4))
for hyd, group in summary.groupby('Hydrology'):
    ax.scatter(group['MeanCO2_kt']/1000, group['MeanTotalCost_MUSD'], label=hyd, s=70)
    for _, row in group.iterrows():
        ax.annotate(row['Scenario'], (row['MeanCO2_kt']/1000, row['MeanTotalCost_MUSD']), xytext=(4,4), textcoords='offset points')
ax.set_xlabel('Cumulative CO$_2$ [Mt]')
ax.set_ylabel('Mean total system cost [MUSD]')
ax.legend()
ax.grid(alpha=0.25)
plt.show()
```
